## 1. Introduction
### Purpose

This notebook extracts the final analytical view vw_olist_full from the MySQL database used during the project and exports it as a CSV file for downstream analysis in Python.

The extraction step separates data retrieval from the analytical workflow, allowing the main analysis notebook to operate on a reproducible local dataset.

```python
from sqlalchemy import create_engine
import pandas as pd
```

## 2. Database Connection

```python
USER = "root"
PASSWORD = "#########"
HOST = "localhost"
PORT = 3306
DATABASE = "olist_db"

engine = create_engine(f"mysql+pymysql://{USER}:{PASSWORD}@{HOST}:{PORT}/{DATABASE}")
```

## 3. Extract Analytical View

```python
# Query para extrair a view final
query = "SELECT * FROM vw_olist_full;"

# Lendo a view para um DataFrame
df = pd.read_sql(query, engine)

# Exibindo as primeiras linhas
print(df.head())
```

## 4. Export Dataset

```python
output_path = "vw_olist_full.csv"

df.to_csv("vw_olist_full.csv", index=False)

print(f"Dataset exported to: {output_path}")
```
